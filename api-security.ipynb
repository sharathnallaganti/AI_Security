## What is an API?

An API lets two software systems communicate


Example:
- A client sends: “Give me user 42.”
- The API processes the request.
- The API returns user 42 as JSON.

An AI application might use APIs to:

Send prompts to an LLM
Retrieve documents from a vector database
Authenticate users
Execute agent tools
Record security logs

## Essential HTTP methods

-

| Method   | Purpose            | Example               |
| -------- | ------------------ | --------------------- |
| `GET`    | Read information   | Get a user            |
| `POST`   | Create something   | Create a user         |
| `PUT`    | Replace a resource | Replace a user        |
| `PATCH`  | Partially update   | Change a user’s email |
| `DELETE` | Delete something   | Delete a user         |





| Code  | Meaning                            |
| ----- | ---------------------------------- |
| `200` | Request succeeded                  |
| `201` | Resource created                   |
| `400` | Invalid request                    |
| `401` | Authentication required or invalid |
| `403` | Authenticated but not authorized   |
| `404` | Resource not found                 |
| `422` | Request validation failed          |
| `429` | Too many requests                  |
| `500` | Server error                       |

#  Create the API

```python
from fastapi import FastAPI

app = FastAPI(
    title="AI Security API",
    version="1.0.0"
)


@app.get("/")
def home():
    return {"message": "My first secure API"}


@app.get("/health")
def health():
    return {"status": "healthy"}


@app.get("/api-info")
def api_info():
    return {
        "name": "AI Security API",
        "version": "1.0.0"
    }
```

```python
## Cell 3 — Create an API client
```

```python
from fastapi.testclient import TestClient

client = TestClient(app)
```

## Test the endpoints

```python
response = client.get("/health")

print("Status code:", response.status_code)
print("Response:", response.json())
```

## Test all endpoints:

```python
for path in ["/", "/health", "/api-info"]:
    response = client.get(path)

    print("Endpoint:", path)
    print("Status:", response.status_code)
    print("JSON:", response.json())
    print()
```

## Add an automated security test

```python
def test_health_endpoint():
    response = client.get("/health")

    assert response.status_code == 200
    assert response.json() == {"status": "healthy"}

test_health_endpoint()

print("Test passed!")
```
