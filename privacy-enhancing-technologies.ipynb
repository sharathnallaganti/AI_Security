# Privacy Enhancing Technologies 

Privacy-Enhancing Technologies (PETs) are tools and techniques designed to protect personal or sensitive information while it is being stored, transmitted, or used.



**"How can we use data while revealing as little private information as possible?”**



## Common Privacy-Enhancing Technologies




| Technology                | What it does                                                                                        | Simple example                                                 |
| ------------------------- | --------------------------------------------------------------------------------------------------- | -------------------------------------------------------------- |
| **Encryption**            | Makes data unreadable without the correct key                                                       | HTTPS protects data sent to a website                          |
| **Anonymization**         | Removes identifying information                                                                     | Removing names and IDs from research data                      |
| **Pseudonymization**      | Replaces identifying information with fake identifiers                                              | Replacing “John Smith” with “User123”                          |
| **VPN**                   | Encrypts traffic between your device and VPN server and masks your public IP from destination sites | Using a VPN on public Wi-Fi                                    |
| **Tor**                   | Routes traffic through multiple relays to improve anonymity                                         | Privacy-focused web browsing                                   |
| **Differential Privacy**  | Adds controlled statistical noise to protect individuals in datasets                                | Publishing statistics without exposing individual records      |
| **Data Masking**          | Hides portions of sensitive information                                                             | `**** **** **** 1234`                                          |
| **Zero-Knowledge Proofs** | Proves something is true without revealing the underlying secret                                    | Proving eligibility without revealing all personal information |



Tor: Tor (short for The Onion Router) is an open-source software and network that lets users browse the internet and communicate anonymously

## How Tor Works (Onion Routing)

Tor uses a technique called **onion routing**.

### Layered Encryption
Your data is wrapped in multiple layers of encryption — like the layers of an onion.

### Three Relays (Nodes)
Your traffic bounces randomly through three volunteer-run computers called relays:

1. **Guard Relay (Entry)** — Knows your real IP address, but not where you are going.
2. **Middle Relay** — Passes the data along without knowing the source or the destination.
3. **Exit Relay** — Peels off the final layer of encryption and sends the traffic to its destination, without knowing who originally sent it.



### PETs vs Security

There's an important distinction:

Security focuses on protecting systems and information from unauthorized access or modification.

Privacy focuses on controlling the collection, use, disclosure, and identification of information about individuals.

They overlap, but they're not identical.

For an exam, remember:

Privacy-enhancing technologies are technologies and techniques that minimize the exposure of personal information while allowing data to be stored, transmitted, analyzed, or otherwise used.
