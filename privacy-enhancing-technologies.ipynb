# Privacy Enhancing Technologies 

Privacy-Enhancing Technologies (PETs) are tools and techniques designed to protect personal or sensitive information while it is being stored, transmitted, or used.



**"How can we use data while revealing as little private information as possible?”**


## What PETs Are Trying to Accomplish

PETs (Privacy-Enhancing Technologies) generally follow a few core ideas:

- **Minimize the data collected**
- **Protect identities**
- **Limit who can see information**
- **Reduce what can be learned about an individual**

### Example
Imagine a hospital wants researchers to analyze patient statistics. Researchers may need information about age and treatment outcomes, but they don't necessarily need patients' names, addresses, or medical-record numbers. PETs can help separate or protect that identifying information.


## Common Privacy-Enhancing Technologies



| PET                               | Main purpose                                                            | Example                                                             |
| --------------------------------- | ----------------------------------------------------------------------- | ------------------------------------------------------------------- |
| **Encryption**                    | Protect data from unauthorized reading                                  | Encrypting a laptop                                                 |
| **Pseudonymization**              | Replace identifiers                                                     | John Smith → Patient 472                                            |
| **Anonymization**                 | Prevent identification of individuals                                   | Publishing anonymous survey data                                    |
| **Data masking**                  | Hide sensitive portions                                                 | `***-**-1234`                                                       |
| **Tokenization**                  | Replace sensitive values with tokens                                    | Credit card → `TOKEN_8472`                                          |
| **Differential privacy**          | Protect individuals in statistical analysis                             | Adding controlled noise to statistics                               |
| **Zero-knowledge proofs**         | Prove something without revealing the underlying information            | Prove you're over 21 without revealing your birthday                |
| **Homomorphic encryption**        | Perform certain computations while data remains encrypted               | Analyze encrypted financial data                                    |
| **Secure multiparty computation** | Multiple parties jointly compute without revealing their private inputs | Banks analyzing combined fraud patterns                             |
| **Federated learning**            | Train models without centralizing all raw data                          | Phones contributing to an ML model without uploading all local data |




**Tor:** Tor (short for The Onion Router) is an open-source software and network that lets users browse the internet and communicate anonymously

## How Tor Works (Onion Routing)

Tor uses a technique called **onion routing**.

### Layered Encryption
Your data is wrapped in multiple layers of encryption — like the layers of an onion.

### Three Relays (Nodes)
Your traffic bounces randomly through three volunteer-run computers called relays:

1. **Guard Relay (Entry)** — Knows your real IP address, but not where you are going.
2. **Middle Relay** — Passes the data along without knowing the source or the destination.
3. **Exit Relay** — Peels off the final layer of encryption and sends the traffic to its destination, without knowing who originally sent it.



### PETs vs Security

There's an important distinction:

Security focuses on protecting systems and information from unauthorized access or modification.

Privacy focuses on controlling the collection, use, disclosure, and identification of information about individuals.

They overlap, but they're not identical.

**For an exam, remember:**

Privacy-enhancing technologies are technologies and techniques that minimize the exposure of personal information while allowing data to be stored, transmitted, analyzed, or otherwise used.
