# Jailbreaking vs. Prompt Injection

## Jailbreaking

**Jailbreaking** is an attempt to bypass an AI model's safety rules, ethical restrictions, or usage policies. The attacker creates prompts intended to make the model perform an action it would normally refuse.

$$
\text{Jailbreaking} =
\text{Attempt to bypass AI safety guardrails}
$$

**Example:** Asking an AI to adopt an unrestricted identity so that it generates prohibited content.

---

## Prompt Injection

**Prompt injection** is an attack technique in which malicious instructions are inserted into the information processed by an AI system. The goal is to override the system's original instructions and make it follow the attacker's commands.

$$
\text{Prompt Injection} =
\text{Malicious input that overrides intended instructions}
$$

Prompt injection can occur in two forms:

1. **Direct prompt injection:** The attacker sends malicious instructions directly to the AI.
2. **Indirect prompt injection:** The instructions are hidden inside external content, such as a webpage, email, or document.

---

## Main Difference

| Jailbreaking | Prompt Injection |
|---|---|
| Targets the model's safety restrictions | Targets the application's instructions or workflow |
| Usually comes directly from the user | Can be direct or hidden in external content |
| Attempts to produce restricted content | Attempts to control the AI's behavior or actions |

$$
\boxed{
\text{Jailbreaking targets safety controls, while prompt injection targets instruction control.}
}
$$

Jailbreaking can be considered a specialized form of direct prompt manipulation, but not every prompt-injection attack is a jailbreak.
