# Threat Modeling for Large Language Model Applications

## 1. Introduction

Threat modeling is a structured process used to identify what must be protected, who may attack the system, how an attack could occur, and which security controls are needed.

For an LLM application, threat modeling must cover more than the language model. It must examine the entire application, including prompts, users, data sources, retrieval systems, tools, APIs, memory, and infrastructure.

$$
\text{Threat Modeling}
=
\text{Identify Assets}
+
\text{Identify Threats}
+
\text{Evaluate Risk}
+
\text{Select Controls}
$$

The main objective is to answer four questions:

1. What are we protecting?
2. Who or what could attack it?
3. How could the attack occur?
4. What controls can reduce the risk?

---

## 2. Threats, Vulnerabilities, and Risks

These terms are related but have different meanings.

### Threat

A **threat** is something capable of causing harm.

Examples include:

- A malicious user
- A compromised employee account
- A poisoned document
- A vulnerable plugin
- A malicious external website
- An attacker-controlled AI agent

### Vulnerability

A **vulnerability** is a weakness that a threat can exploit.

Examples include:

- Missing access controls
- Excessive tool permissions
- Unfiltered model output
- Sensitive information in a system prompt
- Inadequate logging
- Failure to separate instructions from data

### Risk

A **risk** is the possible harm caused when a threat exploits a vulnerability.

$$
\text{Risk}
=
\text{Threat}
\times
\text{Vulnerability}
\times
\text{Impact}
$$

For example:

- **Threat:** A malicious customer
- **Vulnerability:** Weak document-access controls
- **Attack:** The customer manipulates the LLM into retrieving another user's file
- **Impact:** Confidential information is exposed
- **Risk:** Unauthorized cross-user data disclosure

---

## 3. Define the System

Threat modeling begins by documenting the system being evaluated.

The description should identify:

- The LLM provider and model
- The application using the model
- System and developer prompts
- User-input channels
- Databases and document repositories
- Retrieval-Augmented Generation systems
- Embedding models and vector databases
- External APIs and tools
- Authentication systems
- Authorization rules
- Conversation memory
- Logging and monitoring
- Cloud infrastructure
- Human approval processes

A model name alone is not a complete system description.

$$
\text{AI Application}
\neq
\text{LLM Only}
$$

Instead:

$$
\text{AI Application}
=
\text{LLM}
+
\text{Prompts}
+
\text{Data}
+
\text{Tools}
+
\text{Users}
+
\text{Infrastructure}
$$

---

## 4. Establish Security Objectives

Security objectives define what the system must protect.

The traditional cybersecurity objectives are confidentiality, integrity, and availability.

### Confidentiality

Information should be accessible only to authorized users.

Examples include:

- Protecting customer information
- Preventing system-prompt leakage
- Securing API keys
- Separating data belonging to different organizations
- Preventing conversation-history exposure

### Integrity

Information and system behavior should not be altered without authorization.

Examples include:

- Preventing knowledge-base poisoning
- Protecting system prompts from modification
- Preventing unauthorized tool commands
- Detecting altered model configurations
- Preserving accurate audit records

### Availability

The system should remain available to authorized users.

Examples include:

- Preventing excessive token consumption
- Limiting recursive tool calls
- Blocking denial-of-service prompts
- Applying rate limits
- Controlling expensive model requests

$$
\text{Security Objectives}
=
\text{Confidentiality}
+
\text{Integrity}
+
\text{Availability}
$$

LLM systems may require additional objectives:

- Privacy
- Safety
- Accountability
- Transparency
- Reliability
- Human control
- Regulatory compliance

## 5. Identify Valuable Assets

An asset is anything that would cause harm to the organization if it were exposed, modified, destroyed, or misused.

### Data Assets

- Customer records
- Employee information
- Financial records
- Medical information
- Authentication credentials
- Conversation histories
- Internal documents
- Training datasets
- Retrieval documents
- Model outputs

### AI Assets

- Model weights
- System prompts
- Prompt templates
- Fine-tuning datasets
- Embedding models
- Safety configurations
- Evaluation datasets
- Model-routing rules

### Technical Assets

- API keys
- Access tokens
- Cloud credentials
- Databases
- Vector stores
- Source code
- Application logs
- Tool configurations
- Deployment pipelines

### Business Assets

- Organizational reputation
- Customer trust
- Intellectual property
- Regulatory compliance
- Business operations
- Decision-making processes

Assets should be classified according to sensitivity.

| Classification | Description | Example |
|---|---|---|
| Public | Information approved for public release | Published documentation |
| Internal | Information intended only for employees | Internal procedures |
| Confidential | Information requiring controlled access | Customer records |
| Restricted | Highly sensitive information | Passwords and private keys |

## 6. Identify Threat Actors

A threat actor is a person, organization, system, or process that may cause harm.

### External Attackers

External attackers may attempt to:

- Steal sensitive information
- Manipulate model behavior
- Abuse connected tools
- Damage organizational reputation
- Increase operating costs
- Disrupt system availability

### Malicious Users

An authorized user may attempt to exceed their permitted access.

Examples include:

- A customer trying to access another customer's information
- A student attempting to obtain restricted examination material
- An employee asking the model for confidential management documents

### Insider Threats

Employees or contractors may misuse legitimate access.

Insider activity may include:

- Uploading poisoned documents
- Changing system prompts
- Copying model logs
- Misusing administrative tools
- Disabling security monitoring

### Compromised Third Parties

An external service may be compromised and return malicious content to the AI application.

Examples include:

- Search-engine results containing prompt injections
- Compromised plugins
- Malicious API responses
- Poisoned software packages
- Manipulated external documents

### Unintentional Threat Actors

Not all harm is intentional.

A user may accidentally:

- Upload confidential information
- Approve an unsafe tool action
- rely on an incorrect model response
- Share generated content containing private data
- Misconfigure application permissions

## 7. Create a Data-Flow Diagram

A data-flow diagram shows how information moves through the system.

A simplified LLM application may follow this flow:

$$
\text{User}
\rightarrow
\text{Application}
\rightarrow
\text{LLM}
\rightarrow
\text{Response}
$$

A RAG-based AI agent may have a more complex flow:

$$
\text{User Input}
\rightarrow
\text{Application}
\rightarrow
\text{Retriever}
\rightarrow
\text{Vector Database}
$$

$$
\text{Retrieved Documents}
\rightarrow
\text{LLM}
\rightarrow
\text{Tool Call}
\rightarrow
\text{External System}
$$

The diagram should identify:

- Processes
- Data stores
- External entities
- Communication paths
- Security controls
- Trust boundaries

## 8. Identify Trust Boundaries

A trust boundary exists where data moves between components with different trust levels or permissions.

Common trust boundaries include:

- User input entering the application
- Retrieved documents entering the model context
- Model output entering another application
- Tool requests leaving the AI system
- Third-party API responses entering the system
- Data moving between organizational tenants
- Administrative interfaces connecting to production systems

$$
\text{Crossing a Trust Boundary}
\Rightarrow
\text{Validation and Authorization Required}
$$

A fundamental security rule is:

$$
\boxed{
\text{Retrieved Content Is Data, Not a Trusted Instruction}
}
$$

For example, a webpage may contain a malicious instruction telling the AI to send confidential information to an attacker. The application should prevent the model from treating that webpage instruction as authorized.

## 9. Identify the Attack Surface

The attack surface includes every location where an attacker can interact with or influence the system.

### User-Facing Attack Surface

- Chat interfaces
- Search boxes
- Uploaded files
- Images
- Audio input
- API requests
- Conversation history

### Data Attack Surface

- Training data
- Fine-tuning data
- Retrieval documents
- Vector databases
- User profiles
- Application memory
- Cached model responses

### Integration Attack Surface

- Plugins
- External APIs
- Email systems
- Browsers
- Code interpreters
- Databases
- Cloud services
- Payment systems

### Infrastructure Attack Surface

- Model endpoints
- Containers
- Kubernetes clusters
- Cloud storage
- Identity systems
- Logging services
- Software dependencies
- Deployment pipelines

The attack surface grows when the model receives more permissions or connects to additional tools.

$$
\text{Attack Surface}
\propto
\text{Data Access}
+
\text{Tool Access}
+
\text{Autonomy}
$$

## 10. Identify LLM-Specific Threats

### Prompt Injection

An attacker provides instructions intended to override the application's original task.

### Jailbreaking

An attacker attempts to bypass the model's safety controls and produce prohibited behavior.

### Sensitive-Information Disclosure

The model reveals information that should remain confidential.

### Data and Model Poisoning

An attacker modifies training, fine-tuning, or retrieval data to influence model behavior.

### Excessive Agency

The AI receives more permissions or autonomy than required.

### Insecure Output Handling

The application executes or renders unsafe model output without validation.

### Model Denial of Service

An attacker causes excessive resource consumption through long prompts, repeated requests, or recursive tool calls.

### Supply-Chain Vulnerabilities

A compromised model, dataset, library, plugin, or external service affects the application.

### Misinformation

The system produces inaccurate information that users may incorrectly trust.

### Model Theft

An attacker attempts to copy model behavior, parameters, or proprietary knowledge.

## 11. Applying STRIDE to an LLM Application

STRIDE is a traditional threat-modeling framework that can also be applied to AI systems.

| STRIDE category | Meaning | LLM example |
|---|---|---|
| Spoofing | Pretending to be another identity | Using a stolen account to access an AI assistant |
| Tampering | Unauthorized modification | Poisoning documents in a RAG knowledge base |
| Repudiation | Denying an action occurred | Missing logs make AI tool actions difficult to prove |
| Information disclosure | Unauthorized data exposure | The model reveals another user's records |
| Denial of service | Reducing system availability | Long prompts consume excessive resources |
| Elevation of privilege | Gaining unauthorized permissions | A user manipulates an agent into using an administrator tool |

STRIDE helps organize technical threats, but it should be supplemented with AI-specific risks such as hallucination, prompt injection, model poisoning, and unsafe autonomy.

## 12. Develop Abuse Cases

An abuse case describes how an attacker may misuse the system.

A useful abuse-case format is:

> As an attacker, I want to exploit a weakness so that I can cause a specific harmful result.

### Example 1: Data Leakage

> As a malicious customer, I want to manipulate the chatbot into retrieving another customer's records so that I can obtain confidential information.

### Example 2: Indirect Prompt Injection

> As an attacker, I want to place hidden instructions on a webpage so that the AI assistant follows my instructions when it summarizes the page.

### Example 3: Excessive Agency

> As a compromised user, I want the AI agent to approve a sensitive transaction without human confirmation.

### Example 4: Knowledge-Base Poisoning

> As a malicious employee, I want to upload a manipulated document so that the model gives users false security instructions.

## 13. Build an Attack Tree

An attack tree starts with an attacker's objective and divides it into possible attack paths.

### Goal: Obtain Confidential Customer Data

Possible paths include:

1. Manipulate the model through prompt injection
2. Exploit missing document-access controls
3. Retrieve another user's conversation
4. Cause the agent to call an administrative tool
5. Extract sensitive information from logs
6. Use a compromised plugin
7. Abuse an exposed API key

The attack succeeds if any practical path reaches the target asset.

$$
\text{Attack Goal}
=
P_1 \lor P_2 \lor P_3 \lor \cdots \lor P_n
$$

For a multi-step attack requiring every stage:

$$
\text{Successful Attack}
=
P_1 \land P_2 \land P_3
$$

Attack trees help investigators identify which controls interrupt the greatest number of attack paths.

## 14. Evaluate Existing Controls

After identifying threats, the organization should document existing safeguards.

### Preventive Controls

- Authentication
- Role-based access control
- Prompt-injection filtering
- Tool allowlists
- Network restrictions
- Input validation
- Output sanitization
- Rate limiting
- Human approval

### Detective Controls

- Prompt logging
- Tool-call logging
- Data-access monitoring
- Anomaly detection
- Security alerts
- Model-response monitoring
- Canary tokens

### Corrective Controls

- Session termination
- Credential revocation
- Tool disabling
- Model rollback
- Knowledge-base restoration
- Incident-response procedures
- User notification

Defense in depth uses multiple controls:

$$
\text{Defense in Depth}
=
\text{Prevent}
+
\text{Detect}
+
\text{Respond}
+
\text{Recover}
$$

## 15. Score and Prioritize Risks

Threats should be prioritized according to likelihood and impact.

A simple scoring model uses values from 1 to 5:

$$
\text{Risk Score}
=
\text{Likelihood}
\times
\text{Impact}
$$

| Score | Risk level | Recommended response |
|---:|---|---|
| 20–25 | Critical | Stop deployment until mitigated |
| 12–19 | High | Correct before production use |
| 6–11 | Medium | Mitigate and monitor |
| 1–5 | Low | Document and review periodically |

### Example Risk Register

| Threat | Likelihood | Impact | Score | Risk level |
|---|---:|---:|---:|---|
| Cross-user data leakage | 4 | 5 | 20 | Critical |
| Indirect prompt injection | 4 | 4 | 16 | High |
| Excessive token usage | 3 | 3 | 9 | Medium |
| System-prompt disclosure | 3 | 2 | 6 | Medium |

The rating should also consider:

- Ease of exploitation
- Number of affected users
- Data sensitivity
- Required attacker access
- Availability of detection
- Legal consequences
- Financial damage
- Reputational harm

## 16. Example Threat Model

Consider a customer-service AI assistant that:

- Answers customer questions
- Retrieves customer records
- Searches internal documents
- Creates support tickets
- Sends email responses

### Assets

- Customer records
- Internal support documents
- Email accounts
- API credentials
- Conversation logs

### Entry Points

- Customer chat messages
- Uploaded documents
- Retrieved knowledge-base content
- Email content
- Tool responses

### Major Threats

1. A user accesses another customer's record.
2. A poisoned document instructs the model to reveal information.
3. The model sends an email without approval.
4. Sensitive information appears in application logs.
5. An attacker causes repeated expensive tool calls.
6. The model produces an incorrect support recommendation.

### Recommended Controls

- Enforce record access outside the model
- Filter retrieval results according to user identity
- Treat retrieved documents as untrusted
- Require approval before sending email
- Remove confidential information from logs
- Limit requests and tool calls
- Monitor unusual access patterns
- Record all sensitive actions

## 17. Connect Threats to Security Tests

Every important threat should produce one or more test cases.

| Threat | Security test | Expected result |
|---|---|---|
| Prompt injection | Insert malicious instructions into user input | Application preserves its original task |
| Indirect injection | Place instructions inside a retrieved document | Document is treated as untrusted data |
| Data leakage | Request another user's information | Access is denied |
| Excessive agency | Ask the agent to perform a sensitive action | Human approval is required |
| Output injection | Request executable HTML or commands | Output is sanitized |
| Denial of service | Submit oversized or recursive requests | Limits stop excessive processing |
| Knowledge poisoning | Upload manipulated content | Content is reviewed or quarantined |
| Tool misuse | Request an unauthorized tool call | Authorization blocks the action |

Threat modeling and security testing should be connected:

$$
\text{Threat}
\rightarrow
\text{Security Requirement}
\rightarrow
\text{Test Case}
\rightarrow
\text{Evidence}
$$

## 18. Document Residual Risk

Controls may reduce a risk without removing it completely.

$$
\text{Residual Risk}
=
\text{Initial Risk}
-
\text{Control Effectiveness}
$$

For example, prompt filtering may block known attacks but fail against new wording. Human approval may reduce the impact of tool misuse, but users may still approve harmful actions without reviewing them carefully.

Residual risks should be:

- Documented
- Assigned to an owner
- Monitored
- Accepted by authorized leadership
- Reviewed after system changes

---

## 19. Maintain the Threat Model

An LLM threat model must be updated when:

- The model changes
- A new tool is added
- The system receives additional permissions
- New documents are connected
- Authentication rules change
- A new user group is introduced
- The application stores conversation memory
- A vulnerability is discovered
- An incident occurs
- New attack techniques become known

$$
\text{Design Change}
\Rightarrow
\text{Threat Model Review}
\Rightarrow
\text{Security Retesting}
$$

Threat modeling is therefore a continuous activity rather than a one-time document.

## 20. Threat-Modeling Checklist

### System Understanding

- [ ] Model and application versions are documented
- [ ] Data flows are mapped
- [ ] External tools and APIs are identified
- [ ] Trust boundaries are marked
- [ ] User roles are documented

### Asset Identification

- [ ] Sensitive data is classified
- [ ] Credentials and secrets are identified
- [ ] Business-critical operations are documented
- [ ] Model and prompt assets are recorded

### Threat Identification

- [ ] Prompt injection is considered
- [ ] Jailbreaking is considered
- [ ] Data leakage is considered
- [ ] RAG poisoning is considered
- [ ] Tool misuse is considered
- [ ] Excessive agency is considered
- [ ] Denial of service is considered
- [ ] Supply-chain risk is considered

### Control Review

- [ ] Authentication is enforced
- [ ] Authorization is enforced outside the model
- [ ] Tool permissions follow least privilege
- [ ] Sensitive actions require approval
- [ ] Inputs and outputs are validated
- [ ] Security events are logged
- [ ] Rate and resource limits are configured
- [ ] Incident-response procedures exist

### Validation

- [ ] Each high-risk threat has a security test
- [ ] Failed tests are documented
- [ ] Remediation is verified
- [ ] Residual risks are approved
- [ ] The threat model has an assigned owner

## 21. Conclusion

Threat modeling helps an organization identify and address LLM security risks before they become incidents. It connects business assets, technical architecture, attacker behavior, security controls, and testing requirements.

A successful LLM threat model should:

1. Describe the complete AI system
2. Identify valuable assets
3. Map data flows and trust boundaries
4. Identify threat actors and attack paths
5. Evaluate likelihood and impact
6. Recommend preventive and detective controls
7. Convert threats into security test cases
8. Document and monitor residual risks

$$
\boxed{
\text{Effective Threat Modeling}
=
\text{Understand the System}
+
\text{Anticipate Attacks}
+
\text{Prioritize Risks}
+
\text{Validate Controls}
}
$$

The most important principle is that the model should never be the only security control. Authentication, authorization, data isolation, output validation, monitoring, and human approval must be enforced by the surrounding application.
